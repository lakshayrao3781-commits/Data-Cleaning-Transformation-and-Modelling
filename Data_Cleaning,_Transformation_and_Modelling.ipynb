Question 1 : What is Data Loading in Power BI and why is it considered the first step of analysis?

Answer : Data Loading:

Data Loading is the process of importing data from different sources into Microsoft Power BI.

Why it is the first step:

Analysis cannot begin without data.
It helps bring raw data into Power BI for cleaning and visualization.
All dashboards, reports, and models depend on loaded data.

Question 2 : Explain the difference between “Load” and “Transform Data” in Power BI.

Answer : Load:

.Loads data directly into Power BI
.No data cleaning is performed
.Faster process
.Used when data is already clean
.Data goes directly to Data View
.Cannot remove duplicates or change formats
.Simple import process

Transform Data:

.Opens Power Query Editor before loading
.Data cleaning and transformation can be performed
.Takes more time because data is modified
.Used when data contains errors, nulls, or duplicates
.Data is edited first, then loaded
.Can remove duplicates, replace nulls, split columns, change data types, etc.
.Advanced data preparation process

Question 4 : Why is Star Schema preferred over Snowflake Schema in Power BI?

1. Better Performance:
Star Schema uses fewer joins.
Reports load faster.

2. Simpler Model:
Easy to understand and maintain.

3. Better for Visualization:
Works efficiently with Power BI relationships.

4. Easier DAX Calculations:
Measures and KPIs become simpler.

Answer :
Steps in Power Query:

Open Transform Data.
Select columns:
Date
Country
State

Home → Remove Rows → Remove Duplicates
Power BI removes repeated records automatically.

Power Query M Code:
Table.Distinct(Source, {"Date","Country","State"})

Question 6 : Identify and replace null values in
Vaccination_Status.

Answer : Steps:
Open Transform Data.
Select Vaccination_Status column.
Go to:
Transform → Replace Values
Replace:
null or blank
with "Unknown" or "Not Available"

Power Query M Code:
Table.ReplaceValue(
    Source,
    null,
    "Unknown",
    Replacer.ReplaceValue,
    {"Vaccination_Status"}
)

Question 8 : Create a summarized table showing total confirmed cases by Country.

Answer: DAX Formula:

Country Summary =
SUMMARIZE(
    Corona_Virus_2020,
    Corona_Virus_2020[Country],
    "Total Confirmed Cases",
    SUM(Corona_Virus_2020[Confirmed_Cases])
)

Output Example:
Country	Total Confirmed Cases
India	Total Cases
USA	Total Cases
Brazil	Total Cases
Italy	Total Cases

Steps:

Go to Modeling tab.

Click New Table.

Paste the DAX formula.

Press Enter.
